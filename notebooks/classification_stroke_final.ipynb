## Section B — Preprocessing & Feature Engineering

### B1 — Data cleaning
- Drop the single `gender == 'Other'` row (too sparse — 1 out of 5,110 — to let a model learn anything from it, and it would create a category unseen in a stratified split anyway).
- Impute missing `bmi` values (median imputation, robust to the right-skew/outliers seen in Section A2), fitted on the training set only in B2.
- Check and treat `bmi` outliers using the IQR method.
- Duplicates: none were found in A1, so no action needed there.

```python
print("Rows before dropping 'Other' gender:", len(df))
df = df[df['gender'] != 'Other'].reset_index(drop=True)
print("Rows after:", len(df))
print(df['gender'].value_counts())
```

```python
# Outlier check on bmi using the IQR method
Q1 = df['bmi'].quantile(0.25)
Q3 = df['bmi'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['bmi'] < lower_bound) | (df['bmi'] > upper_bound)]
print(f"BMI IQR bounds: [{lower_bound:.1f}, {upper_bound:.1f}]")
print(f"Number of bmi outliers: {len(outliers)} ({len(outliers)/len(df)*100:.1f}% of rows)")
```

```python
# Cap (winsorize) bmi outliers to the IQR bounds rather than dropping rows,
# since stroke is already rare and we don't want to lose positive cases living in the tails
df['bmi'] = df['bmi'].clip(lower=lower_bound, upper=upper_bound)
print(df['bmi'].describe())
```

**Observation (B1):** Clipping (winsorizing) rather than dropping outlier rows was chosen because several of the high-BMI outliers are stroke-positive cases (rare, high-value data points) — deleting them would worsen the already severe 4.87% class imbalance. Missing `bmi` values are deferred to the pipeline's imputer in B2 so imputation statistics are learned from the training fold only, avoiding leakage.

### B2 — Encoding, scaling & splitting
- Stratified 80:20 train/test split on `stroke` (preserves the 4.87% positive rate in both sets).
- `SimpleImputer` (median) for `bmi`, fitted on the training set only.
- One-hot encoding for nominal categoricals (`work_type`, `smoking_status`, `Residence_type`) and binary mapping for `gender`/`ever_married`.
- `StandardScaler` for numeric features, fitted on the training set only, then applied to both train and test.
- All of this is wrapped in a scikit-learn `ColumnTransformer` + `Pipeline` so the fit-on-train-only rule is enforced automatically and can't leak into the test set.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

X = df.drop(columns=['id', 'stroke'])
y = df['stroke']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
print("Train stroke rate: %.4f" % y_train.mean())
print("Test stroke rate:  %.4f" % y_test.mean())
```

```python
numeric_features = ['age', 'avg_glucose_level', 'bmi']
binary_features = ['gender', 'ever_married', 'hypertension', 'heart_disease', 'Residence_type']
nominal_features = ['work_type', 'smoking_status']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

binary_transformer = Pipeline(steps=[
    ('ordinal', OrdinalEncoder())
])

nominal_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore', drop='if_binary'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('bin', binary_transformer, binary_features),
    ('nom', nominal_transformer, nominal_features)
])

preprocessor
```

```python
# Fit on the training set ONLY, then transform both splits (no leakage)
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

feature_names = preprocessor.get_feature_names_out()
print("Processed train shape:", X_train_processed.shape)
print("Number of output features:", len(feature_names))
print(list(feature_names))
```

**Observation (B2):** hypertension, heart_disease, and residence type are naturally binary/ordinal so they're ordinal-encoded rather than one-hot, keeping the feature count down; `work_type` and `smoking_status` are true nominal categories (no inherent order) so they get one-hot encoded. The stratified split keeps the train and test stroke rates nearly identical (~4.9% each), which matters a lot given how rare the positive class is.

### B3 — Feature engineering
**Engineered feature: `age_glucose_risk` = `age` × `avg_glucose_level`.**

Justification: Section A's scatter plot (Age vs Avg Glucose Level, by Stroke Outcome) showed stroke cases concentrated where *both* age and glucose are elevated simultaneously, not where either is elevated alone — and the correlation heatmap showed neither feature is strongly linearly related to `stroke` on its own (r ≈ 0.25 and 0.13 respectively). A simple multiplicative interaction term lets linear models like Logistic Regression directly capture that combined-risk pattern instead of relying on the model to discover the interaction implicitly, which tree/kernel-based models can find on their own but linear ones cannot.

```python
df['age_glucose_risk'] = df['age'] * df['avg_glucose_level']

# Compare the new feature's correlation with stroke against its parent features
print(df[['age', 'avg_glucose_level', 'age_glucose_risk', 'stroke']].corr()['stroke'])
```

**Observation (B3):** `age_glucose_risk` correlates with `stroke` more strongly (≈0.27) than either `age` (≈0.25) or `avg_glucose_level` (≈0.13) alone, confirming the interaction adds signal beyond the two original features. This engineered column needs to be added to `X` and re-run through the same preprocessing pipeline (fit on train only) before it's used in Section D's models — done in the cell below.

```python
X = df.drop(columns=['id', 'stroke'])
y = df['stroke']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

numeric_features = ['age', 'avg_glucose_level', 'bmi', 'age_glucose_risk']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('bin', binary_transformer, binary_features),
    ('nom', nominal_transformer, nominal_features)
])

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
feature_names = preprocessor.get_feature_names_out()
print("Final processed train shape:", X_train_processed.shape)
print(list(feature_names))
```

## Section D — Classification Track, Part A

### D1 — Implementation
Train all 5 Part-A algorithms on the same preprocessed train split from Section B and generate predictions on the same test split:

1. Logistic Regression — baseline; interpret coefficients/odds
2. K-Nearest Neighbors — discuss distance metric
3. Naive Bayes (Gaussian) — discuss the conditional independence assumption
4. Decision Tree Classifier — visualise the tree
5. Support Vector Machine (SVC) — features already scaled in Section B

Given the 4.87% stroke prevalence found in Section A, `class_weight='balanced'` is used wherever the estimator supports it (Logistic Regression, Decision Tree, SVC) so the models aren't just predicting "no stroke" for everyone. KNN and Gaussian Naive Bayes have no direct `class_weight` equivalent — this is called out as a limitation below and is one reason ensemble methods (Part B, Review 2) tend to handle imbalance better.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              confusion_matrix, roc_auc_score, ConfusionMatrixDisplay,
                              RocCurveDisplay)

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, class_weight='balanced',
                                               random_state=RANDOM_STATE),
    'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2),
    'Naive Bayes (Gaussian)': GaussianNB(),
    'Decision Tree': DecisionTreeClassifier(max_depth=5, class_weight='balanced',
                                             random_state=RANDOM_STATE),
    'SVM (SVC)': SVC(kernel='rbf', C=1.0, probability=True, class_weight='balanced',
                      random_state=RANDOM_STATE)
}

fitted_models = {}
predictions = {}
probabilities = {}

for name, model in models.items():
    model.fit(X_train_processed, y_train)
    fitted_models[name] = model
    predictions[name] = model.predict(X_test_processed)
    probabilities[name] = model.predict_proba(X_test_processed)[:, 1]
    print(f"{name}: trained, {len(predictions[name])} predictions generated, no errors.")
```

**Observation (D1):** All 5 models trained and produced predictions on the held-out test set without error. The next step evaluates each with the mandatory metric set — Accuracy, Precision, Recall, weighted F1, Confusion Matrix, and ROC-AUC — before assembling the preliminary comparison table.

### D2 — Evaluation
Accuracy, Precision, Recall, weighted F1, and ROC-AUC per algorithm, plus each confusion matrix, then a single preliminary comparison table.

```python
results = []
for name in models:
    y_pred = predictions[name]
    y_proba = probabilities[name]
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, zero_division=0),
        'Recall': recall_score(y_test, y_pred),
        'F1 (weighted)': f1_score(y_test, y_pred, average='weighted'),
        'ROC-AUC': roc_auc_score(y_test, y_proba)
    })

results_df = pd.DataFrame(results).set_index('Model').round(4)
results_df = results_df.sort_values('F1 (weighted)', ascending=False)
results_df
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()
for ax, (name, model) in zip(axes, fitted_models.items()):
    cm = confusion_matrix(y_test, predictions[name])
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Stroke', 'Stroke'])
    disp.plot(ax=ax, colorbar=False, cmap='Blues')
    ax.set_title(name)
axes[-1].axis('off')
plt.tight_layout()
plt.show()
```

**Observation (D2):** The results table exposes a real pitfall of this dataset rather than a clean win for any one model. **K-Nearest Neighbors has the highest accuracy (94.7%) and highest weighted F1 (0.925) — but a recall of 0.00: it never predicts a single stroke case.** Because `stroke=1` is only 4.87% of the data, "predict no-stroke for everyone" scores extremely well on accuracy and on *weighted* F1 (which weights by class support and is therefore dominated by the majority class), while being clinically useless. **Naive Bayes sits at the opposite extreme** — recall 0.98 (it flags almost every real stroke case) but precision only 0.074 and accuracy 0.399, because it also flags huge numbers of healthy patients as high-risk. Logistic Regression, SVM, and the Decision Tree — the three models using `class_weight='balanced'` — land in a more usable middle ground: recall between 0.68-0.80 with accuracy around 0.69-0.74. This is exactly why weighted F1 and accuracy are misleading here, and why Precision, Recall, and ROC-AUC (which don't collapse to a single class-weighted number) are the metrics that actually matter for this imbalanced target.

```python
fig, ax = plt.subplots(figsize=(7, 6))
for name, model in fitted_models.items():
    RocCurveDisplay.from_predictions(y_test, probabilities[name], name=name, ax=ax)
ax.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Chance')
ax.set_title('ROC Curves — Part A Classifiers')
ax.legend(loc='lower right', fontsize=8)
plt.tight_layout()
plt.show()
```

**Observation:** Logistic Regression has the highest ROC-AUC (0.838), followed closely by Naive Bayes (0.802), Decision Tree (0.799), and SVM (0.794) — all four separate the classes reasonably well *across thresholds*, even though Naive Bayes and the Decision Tree looked weak on accuracy/precision at the default 0.5 threshold. K-Nearest Neighbors is clearly the worst ranker (ROC-AUC 0.602, barely better than chance), consistent with its recall of 0 above — it isn't just using a bad threshold, it fundamentally isn't separating the classes well. This threshold-independent view is a fairer comparison than the confusion matrices alone, and suggests Naive Bayes and the Decision Tree could become genuinely useful with a lower decision threshold than the default 0.5 — worth revisiting in Review 2.

```python
# Decision tree visualisation (per algorithm note: 'visualise the tree')
fig, ax = plt.subplots(figsize=(20, 8))
plot_tree(fitted_models['Decision Tree'], max_depth=3, feature_names=feature_names,
          class_names=['No Stroke', 'Stroke'], filled=True, fontsize=8, ax=ax)
ax.set_title('Decision Tree Classifier (first 3 levels shown)')
plt.tight_layout()
plt.show()
```

```python
# Logistic Regression coefficients (interpret coefficients/odds, per algorithm note)
coef_df = pd.DataFrame({
    'feature': feature_names,
    'coefficient': fitted_models['Logistic Regression'].coef_[0]
})
coef_df['odds_ratio'] = np.exp(coef_df['coefficient'])
coef_df = coef_df.sort_values('coefficient', key=abs, ascending=False)
coef_df.head(10)
```

**Observation:** The decision tree's top split is on `age`, matching Section A/B's finding that age is the single strongest driver of stroke risk — with `age_glucose_risk` and `avg_glucose_level` appearing in the next levels, validating the Section B3 engineered feature's usefulness. For Logistic Regression, the largest-magnitude coefficients (by odds ratio) also belong to `age`, `age_glucose_risk`, and `hypertension`/`heart_disease` — each positive coefficient here means a one-standard-deviation increase in that (scaled) feature multiplies the odds of stroke by `exp(coefficient)`, holding other features fixed.

## E1 — Implementation (rubric A1): train all 5 Part-B algorithms
First, every algorithm is trained with sensible **default/baseline** settings so the effect of tuning can be measured afterwards.

```python
import warnings
warnings.filterwarnings('ignore')   # silence sklearn convergence/deprecation noise
import sklearn
from sklearn.ensemble import (RandomForestClassifier, AdaBoostClassifier,
                              GradientBoostingClassifier, BaggingClassifier)
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import precision_recall_curve

print("scikit-learn version:", sklearn.__version__, "(MLPClassifier sample_weight support needs >= 1.7)")

# Balanced sample weights for estimators without a class_weight argument
sw_train = compute_sample_weight(class_weight='balanced', y=y_train)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# name -> (baseline estimator, whether to pass sample_weight to fit)
partB_baselines = {
    'Random Forest': (RandomForestClassifier(n_estimators=100, class_weight='balanced_subsample',
                                             random_state=RANDOM_STATE, n_jobs=-1), False),
    'AdaBoost': (AdaBoostClassifier(n_estimators=50, learning_rate=1.0,
                                    random_state=RANDOM_STATE), True),
    'Gradient Boosting': (GradientBoostingClassifier(random_state=RANDOM_STATE), True),
    'Bagging (Decision Tree)': (BaggingClassifier(
                                    estimator=DecisionTreeClassifier(max_depth=5, class_weight='balanced',
                                                                     random_state=RANDOM_STATE),
                                    n_estimators=50, random_state=RANDOM_STATE, n_jobs=-1), False),
    'MLP (Neural Network)': (MLPClassifier(hidden_layer_sizes=(64, 32), activation='relu', max_iter=500,
                                           random_state=RANDOM_STATE), True),
}

def fit_model(model, uses_sw, X, y, sw=None):
    """Fit with balanced sample weights when the estimator needs them."""
    return model.fit(X, y, sample_weight=sw) if uses_sw else model.fit(X, y)

base_models, base_pred, base_proba = {}, {}, {}
for name, (est, uses_sw) in partB_baselines.items():
    m = fit_model(est, uses_sw, X_train_processed, y_train, sw_train)
    base_models[name] = m
    base_pred[name] = m.predict(X_test_processed)
    base_proba[name] = m.predict_proba(X_test_processed)[:, 1]
    print(f"{name}: trained, {len(base_pred[name])} test predictions generated, no errors.")
```

**Observation (E1):** All five Part-B models trained on the same preprocessed train split and produced predictions on the same 1,022-row test set (50 stroke cases) without error. These are *baseline* settings. Even before tuning, the effect of imbalance handling is visible: AdaBoost (recall 0.80) and Bagging (0.80) already flag most strokes, while the **un-tuned Random Forest never predicts a stroke (recall 0.00)** and the un-tuned MLP catches only 24% of them, despite 94.8% / 88.9% accuracy. That repeats the Part-A lesson that accuracy is not the metric to trust on this target.

## E2 — Hyperparameter tuning (rubric A3)
Stratified 5-fold `GridSearchCV` (scoring = ROC-AUC) on the training set for each Part-B algorithm, using the parameters named in the guidelines (plus a few that strongly affect behaviour on imbalanced data).

```python
param_grids = {
    'Random Forest': {'n_estimators': [100, 300], 'max_depth': [4, 8, None], 'min_samples_leaf': [1, 5]},
    'AdaBoost': {'n_estimators': [50, 100, 200, 300], 'learning_rate': [0.01, 0.1, 0.5, 1.0]},
    'Gradient Boosting': {'n_estimators': [100, 200], 'learning_rate': [0.05, 0.1],
                          'max_depth': [2, 3]},
    'Bagging (Decision Tree)': {'n_estimators': [50, 100], 'max_samples': [0.5, 1.0],
                                'estimator__max_depth': [3, 5, 8]},
    'MLP (Neural Network)': {'hidden_layer_sizes': [(32,), (64,), (64, 32), (128, 64)],
                             'activation': ['relu', 'tanh', 'logistic']},
}

tuned_models, tuned_pred, tuned_proba, grid_results = {}, {}, {}, {}
for name, (est, uses_sw) in partB_baselines.items():
    gs = GridSearchCV(est, param_grids[name], scoring='roc_auc', cv=cv, n_jobs=-1, refit=True)
    if uses_sw:
        gs.fit(X_train_processed, y_train, sample_weight=sw_train)
    else:
        gs.fit(X_train_processed, y_train)
    grid_results[name] = gs
    tuned_models[name] = gs.best_estimator_
    tuned_pred[name] = gs.predict(X_test_processed)
    tuned_proba[name] = gs.predict_proba(X_test_processed)[:, 1]
    print(f"{name}: best CV ROC-AUC = {gs.best_score_:.4f} | best params = {gs.best_params_}")
```

**Observation (E2 – best parameters):** The grid searches (ROC-AUC, 5-fold CV on the training set) selected: Random Forest `n_estimators=300, max_depth=None, min_samples_leaf=5` (CV 0.8305); AdaBoost `n_estimators=200, learning_rate=0.5` (0.8384); Gradient Boosting `n_estimators=100, learning_rate=0.05, max_depth=2` (0.8279); Bagging `n_estimators=100, max_samples=0.5` with a shallow `max_depth=3` base tree (0.8276); MLP `hidden_layer_sizes=(64,)` with `activation='logistic'` (0.8380). The pattern is consistent: for this noisy, imbalanced target the winners are the **simpler / more regularised configurations** (shallow trees, a low learning rate, a single hidden layer), because deeper or larger models overfit the 199 training stroke cases. (The MLP printed convergence warnings at `max_iter=500`; the selected network is still the best by CV, but it could be trained longer.)

```python
# Baseline vs tuned (held-out test set) for every Part-B algorithm
def metric_row(y_true, y_pred, y_proba):
    return {'Accuracy': accuracy_score(y_true, y_pred),
            'Precision': precision_score(y_true, y_pred, zero_division=0),
            'Recall': recall_score(y_true, y_pred),
            'F1 (weighted)': f1_score(y_true, y_pred, average='weighted'),
            'F1 (stroke class)': f1_score(y_true, y_pred),
            'ROC-AUC': roc_auc_score(y_true, y_proba)}

rows = []
for name in partB_baselines:
    for tag, pred, proba in [('baseline', base_pred, base_proba), ('tuned', tuned_pred, tuned_proba)]:
        r = {'Model': name, 'Version': tag}
        r.update(metric_row(y_test, pred[name], proba[name]))
        rows.append(r)
tuning_df = pd.DataFrame(rows).set_index(['Model', 'Version']).round(4)
tuning_df
```

```python
# Test ROC-AUC / Recall gain from tuning
gain = pd.DataFrame({
    'ROC-AUC baseline': [tuning_df.loc[(n, 'baseline'), 'ROC-AUC'] for n in partB_baselines],
    'ROC-AUC tuned':    [tuning_df.loc[(n, 'tuned'), 'ROC-AUC'] for n in partB_baselines],
    'Recall baseline':  [tuning_df.loc[(n, 'baseline'), 'Recall'] for n in partB_baselines],
    'Recall tuned':     [tuning_df.loc[(n, 'tuned'), 'Recall'] for n in partB_baselines],
}, index=list(partB_baselines))
gain['ROC-AUC change'] = (gain['ROC-AUC tuned'] - gain['ROC-AUC baseline']).round(4)
gain['Recall change'] = (gain['Recall tuned'] - gain['Recall baseline']).round(4)
gain
```

**Observation (E2 – effect of tuning):** Tuning improved the test ROC-AUC of every Part-B model: Random Forest 0.781→0.826, AdaBoost 0.824→0.838, Gradient Boosting 0.810→0.837, Bagging 0.836→0.839 and **MLP 0.757→0.841 (the biggest gain, +0.084)**. Recall also rose for all five models (MLP 0.24→0.80, Gradient Boosting 0.66→0.84, Random Forest 0.00→0.26, AdaBoost and Bagging 0.80→0.84) at the cost of lower accuracy and weighted F1 – the expected trade-off, since accuracy was inflated by predicting 'no stroke' too often. Because the test set has only 50 stroke cases, each patient is worth 0.02 recall, so differences of ±0.04 are two patients and should not be over-read; the ROC-AUC gains are a steadier indicator. Improvement in at least one metric is therefore documented for all five algorithms.

## E5 — Final model selection (rubric A3)
**Selection rule:** the model with the highest **cross-validated ROC-AUC on the training set** (threshold-independent, not tuned on the test set, and robust to the 4.87% prevalence). Test-set precision/recall are then used to check that the winner is also practically useful as a screening tool. Because the best model is picked by a score that is itself maximised over a grid, its CV score is slightly optimistic — the held-out test score is the unbiased check.

```python
ranking = final_table.sort_values('CV ROC-AUC (train)', ascending=False)[['Model', 'Part', 'CV ROC-AUC (train)', 'ROC-AUC', 'Recall', 'Precision']]
print("Ranking by CV ROC-AUC (train):")
display(ranking)
best_name = ranking.iloc[0]['Model']
print("\nSelected final model:", best_name)
```

**Observation (selection):** By cross-validated ROC-AUC the top three are **Logistic Regression (0.8385), AdaBoost (0.8384) and MLP (0.8380)**. A spread of 0.0005 is far smaller than fold-to-fold noise, so these are statistically tied; the stated rule picks Logistic Regression, and the tie is broken in its favour for practical reasons: it is the simplest model, it is interpretable (odds ratios were examined in Part A and age / `age_glucose_risk` dominate), it produces well-behaved probabilities, trains instantly, and its test recall (0.80) and precision (0.13) are equal or better than the tied alternatives. Choosing a complex model would add no measurable accuracy and make the viva explanation harder.

```python
# --- Step 1: hyperparameter tuning of the selected model (if it was not already tuned in E2) ---
final_grids = {
    'Logistic Regression': {'C': [0.001, 0.01, 0.1, 1, 10, 100], 'penalty': ['l1', 'l2'], 'solver': ['liblinear']},
}
if best_name in partB_baselines:                      # Part-B winner: already tuned in E2
    final_search = grid_results[best_name]
    final_base_model = base_models[best_name]
    uses_sw_final = partB_baselines[best_name][1]
else:                                                 # Part-A winner: tune it now
    final_search = GridSearchCV(models[best_name].__class__(**{**models[best_name].get_params(), 'solver': 'liblinear'}),
                                final_grids[best_name], scoring='roc_auc', cv=cv, n_jobs=-1)
    final_search.fit(X_train_processed, y_train)
    final_base_model = models[best_name]
    uses_sw_final = False
final_model = final_search.best_estimator_
print(f"{best_name}: best params = {final_search.best_params_} | best CV ROC-AUC = {final_search.best_score_:.4f}")

# --- Step 2: decision-threshold tuning on OUT-OF-FOLD training probabilities (test set untouched) ---
oof_proba = cross_val_predict(final_model, X_train_processed, y_train, cv=cv, method='predict_proba',
                              **({'params': {'sample_weight': sw_train}} if uses_sw_final else {}))[:, 1]
prec, rec, thr = precision_recall_curve(y_train, oof_proba)
def best_threshold(beta):
    f = (1 + beta**2) * prec[:-1] * rec[:-1] / np.maximum(beta**2 * prec[:-1] + rec[:-1], 1e-12)
    return thr[int(np.argmax(f))]
thr_f1, thr_f2 = best_threshold(1), best_threshold(2)
print(f"Threshold maximising F1 (stroke class) on OOF train: {thr_f1:.3f}")
print(f"Threshold maximising F2 (recall-weighted) on OOF train: {thr_f2:.3f}")

# --- Step 3: before / after comparison on the held-out test set ---
p_base = final_base_model.predict_proba(X_test_processed)[:, 1]
p_tuned = final_model.predict_proba(X_test_processed)[:, 1]
versions = {
    f'{best_name} - default (Part A, t=0.50)': (p_base, 0.5),
    f'{best_name} - tuned hyperparameters (t=0.50)': (p_tuned, 0.5),
    f'{best_name} - tuned + F1 threshold (t={thr_f1:.2f})': (p_tuned, thr_f1),
    f'{best_name} - tuned + F2 threshold (t={thr_f2:.2f})': (p_tuned, thr_f2),
}
rows = []
for label, (proba, t) in versions.items():
    pred = (proba >= t).astype(int)
    r = {'Version': label}; r.update(metric_row(y_test, pred, proba)); rows.append(r)
final_cmp = pd.DataFrame(rows).set_index('Version').round(4)
display(final_cmp)

# --- Step 4: threshold curve + confusion matrices for the final model ---
fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))
axes[0].plot(thr, prec[:-1], label='Precision (OOF train)', color=palette[0])
axes[0].plot(thr, rec[:-1], label='Recall (OOF train)', color=palette[3])
for t, lab in [(0.5, 'default 0.50'), (thr_f2, f'F2 t={thr_f2:.2f}'), (thr_f1, f'F1 t={thr_f1:.2f}')]:
    axes[0].axvline(t, linestyle='--', color='gray'); axes[0].text(t, 0.02, lab, rotation=90, fontsize=8)
axes[0].set_title('Precision / Recall vs decision threshold'); axes[0].set_xlabel('Threshold'); axes[0].legend()
for ax, (t, ttl) in zip(axes[1:], [(0.5, 'Tuned LR, t = 0.50'), (thr_f2, f'Tuned LR, F2 threshold t = {thr_f2:.2f}  (FINAL)')]):
    cm = confusion_matrix(y_test, (p_tuned >= t).astype(int))
    ConfusionMatrixDisplay(cm, display_labels=['No Stroke', 'Stroke']).plot(ax=ax, colorbar=False, cmap='Blues')
    ax.set_title(ttl)
plt.tight_layout(); plt.show()

# Final model = tuned model + F2 decision threshold
FINAL_THRESHOLD = thr_f2
final_pred = (p_tuned >= FINAL_THRESHOLD).astype(int)
```

**Observation (final model & documented improvement):** Hyperparameter tuning of Logistic Regression chose `C=0.01` with an `l1` penalty (stronger regularisation than the default `C=1`, which also performs feature selection), raising the cross-validated ROC-AUC from 0.8385 to **0.8413** and the test ROC-AUC from 0.8378 to **0.8387**; at the default 0.5 threshold it also lifts recall from 0.80 to 0.84 (precision falls from 0.133 to 0.122). Because a 0.5 cutoff is arbitrary for a 4.9%-prevalence outcome, the decision threshold was then tuned on *out-of-fold training* probabilities only. The **F2-optimal threshold (≈0.64)**, which weights recall twice as heavily as precision (appropriate for a screening task where missing a stroke is costlier than a false alarm), was chosen as the final operating point. Versus the Part-A default model on the test set it improves **precision 0.133→0.182, stroke-class F1 0.227→0.293, accuracy 0.734→0.825 and weighted F1 0.809→0.870**, while recall falls from 0.80 to 0.74 (3 fewer of the 50 strokes caught). The F1-optimal threshold (≈0.75) would raise precision/F1 further (0.25 / 0.35) but drops recall to 0.56, which is too many missed strokes for screening, so it was not chosen. Caveat: with only 50 positive test cases these gains carry real uncertainty; the direction (fewer false alarms for a small recall cost) is more reliable than the exact figures.

**Final model:** Logistic Regression (`C=0.01`, `penalty='l1'`, `class_weight='balanced'`) with a decision threshold of ≈0.64 chosen on out-of-fold training predictions.

**Limitations worth mentioning in the viva**
- The test set has only 50 stroke cases, so single-digit changes in caught strokes move recall by several points.
- The imputer/scaler were fitted on the whole training set before cross-validation (not refitted inside each fold), which can make CV scores very slightly optimistic; the held-out test scores are unaffected.
- The top three models are tied on ROC-AUC; the choice of Logistic Regression rests on simplicity and interpretability, not on a significant accuracy gap.
- The MLP hit `max_iter=500` in some grid configurations (convergence warnings).
